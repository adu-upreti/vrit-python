# Loops

while loop: we can execute a set of statement as long as a condition is true.

```python
a = 1
while a < 5:
    print(a)
    a += 1
```

```python
i = 1
while i < 6:
  print(i)
  if (i == 2):
    break
  i += 1
```

# Continue

```python
a = 0 
while a < 5:
    a += 1
    if a == 3:
        continue
    print(a)    
```

```python
a = 1
while a < 5:
    print(a)
    a += 1

else:
    print("a is no longer less than 5")
```

```python
creat a variable a with value 0
write a while loop that runs as long as a is less than 10
inside the loop : increment a by 2
if a is equals 6, use continue to skip that loop/iteration
print a
```

```python
a = 0
while a < 10:
    a += 2
    if a == 6:
        continue
    print(a)    
```

# For Loops

```python
fruits = ["apple", "mango", "cherry"]

for x in fruits:
    print(x)
```

```python
# Looping though a string
for x in "apple":
    print(x)
```

```python
fruits = ["apple", "mango", "cherry"]

for x in fruits:
    
    if x == "mango":
        break
    print(x)    
```

```python
fruits = ["apple", "mango", "cherry"]

for x in fruits:
    if x == "mango":
        continue
    print(x)    
```

```python
for a in range(5):
    print(a)
```

```python
for a in range(2, 6):
    print(a)
```

```python
for a in range(2, 30, 3):
    print(a)

else:
    print("Loop is finished.")
```

```python
for a in range(2, 30, 3):
    print(a)
    if a == 14: break
    
else:
    print("Loop is finished.")
```

# Nested Loops

```python
color = ["red", "black", "white"]
car = ["mustang", "ford", "tesla"]

for a in color:
    for b in car:
        if  b == "ford": break
        print(a, b)
```

```python
color = ["red", "black", "white"]
car = ["mustang", "ford", "tesla"]

for a in color:
    for b in car:
        print(a, b)
```
