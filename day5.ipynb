# Remove List

```python
alist = ["car", "bus", "jeep", "van"]
alist.remove("bus")
alist.pop(0)
print(alist)
```

```python
alist = ["car", "bus", "jeep", "van"]
alist.pop()
print(alist)
```

```python
alist = ["car", "bus", "jeep", "van"]
del alist[0]
print(alist)
```

```python
alist = ["car", "bus", "jeep", "van"]
alist.clear()
print(alist)
```

# Loop in list

```python
alist = ["car", "bus", "jeep", "van"]
for a in alist:
    print(a)
```

```python
alist = ["car", "bus", "jeep", "van"]
for a in range(len(alist)):
    print(alist[a])
```

```python
atuple = ("red")
atuple1 = ("red",)
print(type(atuple))
print(type(atuple1))
```

# Sets 
- unchangeable but remove and add items
- define between {}
- unindexed
- not allowed duplicates
- unorderd

```python
fruits = {"apple", "banana", "cherry", "apple", 24, True, 1, False, 0}
print(fruits)
print(len(fruits))
```

```python
color = ("red", "green", "yellow", "yellow")
convert = set(color)
print(convert)
```

```python
fruits = {"apples", "banana", "cherry", "apple", 24, 1, False}
for x in fruits:
    print(x)
```

```python
print("apples" not in fruits)
print("apples" in fruits)
```

```python
fruits.add("orange")
print(fruits)
```

```python
name = {"ashish", "lumina", "himal", "orange"}
fruits.update(name)
print(fruits)
```
