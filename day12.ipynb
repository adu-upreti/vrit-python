```python
def table(num):
    for i in range(1, 11):
        print(num, "x", i, "=", num*i)

table(2)
```

```python
def find_largest(numbers):
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest
            
        
nums = [10, 40, 7, 2] 
print(find_largest(nums))
```

# Default Values in __init__()

```python
class Person:
    def __init__(self, name, age=18, city="Pokhara"):
        self.name = name
        self.age = age
        self.city = city

p1 = Person("Rahul")
p2 = Person("Samikshaya", 20, "Kathmandu")

p2.name = "Sulave"
p2.surname = "Karki"

print(p1.name, p1.age)
print(p2.name, p2.age, p2.city, p2.surname)
```

```python
Self parameters links the menthod to the specific object.
```

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(self.name, "says Bhau!")

d1 = Dog("Tommy", 3)
d2 = Dog("Buddy", 5)
d2.bark()
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return "hello " + self.name

    def welcome(self):
        message = self.greet()
        print(message + ", Welcome to our webapp.")

p1 = Person("Aasrya")
p1.welcome()
```
