# Arbitrary Arguments
- the *args parametrs allows a function to accept any numbe rof positional arguments
- *args becomes a tuple inside function

```python
def my_function(greeting, *names):
    for name in names:
        print(greeting, name)
    print(names)

my_function("Namaste", "Sita", "Hari", "Shyam")              
```

```python
def my_function(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(my_function(1, 2, 3))        
```

# Arbitrary Keyword Arguments (**kwargs)
**kwargs parametee allows a function to accept by number of arguments
inside the function kwargs becomes a dictionary 

```python
def my_function(**details):
    print(type(details))
    print("Name:", details['name'])
    print(details)


my_function(name = "Ashish", age = 23, city = "Kathmandu")    
```

```python
def my_function(username, **details):
    print(username)
    print("Additional details:")
    for key, value in details.items():
        print(" ", key + ":", value)


my_function("adi301", age = 23, city = "Kathmandu", hobby = "programming")   
```

```python
def my_function(title, *args, **kwargs):
    print("Title:", title)
    print("positional arguments:", args)
    print("Keyword arguments:", kwargs)

my_function("User Info", "RAM", "Sharma", age = 25, city = "Kathmandu")  
```

```python
The LEGB Rule

Local: Inside the current function.
Global: At the top level of the module/script
Enclosing: inside enclosing functions (from inner to outer)
Built-in: In Python built-in namesapace
```

```python
x = "gloabl"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print("Inner:", x)
    inner()
    print("Outer:", x)

outer()
print("Global:", x)
```
