```python
Question 1: Even Number Counter Function
Create a function count_even(numbers).
Pass a list of numbers.
Use a loop inside the function to count how many numbers are even.
Return the count.
Call the function and print the result.
```

```python
def count_even(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0:
            count += 1
    return count

nums = [1, 2, 3, 4, 5, 6, 8]
result = count_even(nums)
print(result)
```

```python
Question 2: Sum of Odd Numbers
Create a function sum_odd(n).
Use a loop to find the sum of all odd numbers from 1 to n.
Return the sum.
Call the function with any number.
```

```python
def sum_odd(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += 1
    return total
result = sum_odd(10)
print(result)
```
