# Inheritance
- it allows to define a class that inherits all the methods and properties from another class
- Parent class
- Child class

```python
class Pearson:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)

x = Pearson("Jhon", "Khadka")
x.printname()
```

```python
class Student(Pearson):
    pass

x = Student("Sabir", "Shah")
x.printname()
```

# Polymorphism = many forms



```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Drive!")

class Boat:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Sail!")

class Plane:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Fly!")

car1 = Car("Ford", "Mustang")
boat1 = Boat("ABC", "A78d")
plane1 = Plane("Boieng", "A3vh")

for x in (car1, boat1, plane1):
    x.move()
```

# Polymorphism & inheritance 

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Drive!")
class Car(Vehicle):
    pass

class Boat(Vehicle):
    def move(self):
        print("Sail")

class Plane(Vehicle):
    def move(self):
        print("Fly!")

car1 = Car("Ford", "Mustang")
boat1 = Boat("ABC", "A78d")
plane1 = Plane("Boieng", "A3vh")

for x in (car1, boat1, plane1):
    print(x.brand)
    print(x.model)
    x.move()
```

# Encapsulation
- protecting data inside a class

# Why us encapsulation:

- data protection
- validation
- flexibility
- control


```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.__model = model   # private attribute

    def get_brand(self):
        return self.brand      # return brand correctly

    def get_model(self):
        return self.__model

    def solve_puzzle(self, model):
        if model == 2:
            self.__model = model
            return "Model updated successfully"
        else:
            return "Your input is wrong"


p1 = Vehicle("Ford", 20)

print(p1.get_brand())        # Ford
print(p1.get_model())        # 20

print(p1.solve_puzzle(2))    # Update model
print(p1.get_model())        # 2
```
