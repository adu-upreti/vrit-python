# Dictionary 
- ordered
- changeable / mutable
- duplicates
- key value paired

```python
adict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 2000,
    "electric": True
}

print(adict)
print(adict["brand"])
```

```python
print(len(adict))
print(type(adict))
```

```python
adict = dict(name = 'amish', age = 30, grade = 10)
print(adict)
```

```python
x = adict.get("name")
print(x)
keys = adict.keys()
print(keys)
```

```python
adict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 2000,
    "electric": True
}

x = adict.keys()
print(x)
adict['color'] = 'black'
print(adict)
adict['brand'] = 'toyata'
x =adict.values()
print(x)
print(adict.items())
```

```python
if "model" in adict:
    print("Yes, model is define in the adict dictionary.")
```

```python
adict.update({"year": 2010})
print(adict)
```

```python
adict.pop("year")
print(adict)
```

```python
adict.popitem()
print(adict)

del adict["model"]
print(adict)

adict.clear()
print(adict)
```

```python
adict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 2000,
    "electric": True
}

# for acess key of dict using for loop
for a in adict:
    print(a)

# for acess value of dict item 
for a in adict:
    print(adict[a])
```

# Python elif statemnt¶

```python
a = 10
b = 10

if b > a:
    print("b is gretaer than a")

elif a == b:
    print("a and b are equal.")
```

```python
score = int(input("Enter your grade: "))

if score >= 90:
    print("Grade: A")

elif score >= 80:
    print("Grade: B")

elif score >= 70:
    print("Grade: C")

elif score >= 60:
    print("Grade: D")

else:
    print(" Your score is very low, Non-Graded")

```
