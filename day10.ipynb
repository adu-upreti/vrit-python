# JSON

```python
import json

x = '{"name": "Jasmine", "age": 40, "city": "Kathmandu"}'

y = json.loads(x)

print(y["age"])
print(type(y))
```

```python
x = {"name": "Jasmine", "age": 40, "city": "Kathmandu"}

y = json.dumps(x)
print(y)
print(type(y))
```

```python
a = json.dumps(["apple", "bananas"])
print(type(a))
```

# Python Try Expect

```python
try : try block lets you test a blocl of code for erros
except: except block let you handle the the error
else: lets you execute code when there is no error
finally; block lets you execute code, regardless of the result of the try and expect blocks
```

```python
# exception handling
p = "P is defined"
try:
    print(p)
except:
    print(" name 'p' is not defined") 
```

```python
try:
    print(d)
except NameError:
    print("Variable d is not defined")
except:
    print("Smth else went wrong")
```

```python
try:
    print("Hello")
except:
    print("Something went wrong")
else:
    print("Noting went wrong")
```

```python
try:
    print(d)
except NameError:
    print("Variable d is not defined")
finally:
    print("the 'try except' is finished")
```

```python
try:
    f = open("demofile.txt")
    try:
        f.write("Hello")
    except:
        print("Sometime went wrong when writing to the file")
    finally: 
        f.close()
except:
    print("Something went wrong when opening this file")
    
```
