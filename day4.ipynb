# Identity Operators

# is 

```python
x = ["ram", "hari"]
y = ["ram", "hari"]
z = x

#same object as x 
print(x is z) 
# value comparison or x is equal to y
print(x == y)
# x is not the same object as y 
print(x is y)
```

# is not

```python
x = ["ram", "hari"]
y = ["ram", "hari"]
z = x


print(x is not z)
print(x is not y)
print(x != y)
```
