# Global Variables 

```python
x = "global"

def myfunc():
    print("it is " + x + " Variable.")

myfunc()    
```

```python
x = "global"

def myfunc():
    x = "local"
    print("it is " + x + " Variable.")

myfunc()

print("it is " + x + " Variable.")
```

```python
x = "global"

def myfunc():
    global x
    x = "local"

myfunc()

print("it is " + x)
```
